#Glosario

1. División entera (`//`)
- División entre dos numeros la cual su resultado es entero, por lo tanto se redondea al entero menor.
- Ej: 79 // 2 = 39

---
2. Expresión
- Una colección de operadores y números
- Ej: (42 + 7) * 2

---
3. Valor
- Es el resultado de una expresión

---
4. Error: SyntaxError
- Ocurre cuando llamo una función y cometo un error en su escritura
- Ej: abs  42, es un error pues no puse los parentesis en el 42

---
5. Función (`len`)
- Cuenta la cantidad de caracteres que tiene un string. Si no pongo los parentesis tengo un error de sintaxis.

---
6. Error de tipo (TypeError)
- Cuando se ejecuta una operación que no esta asociada a un tipo
- Ej: dividir un string pues con string solo puedo "*" y "+".

---
7. Debugging
- Proces de rastrear y corregir errores (bugs)
